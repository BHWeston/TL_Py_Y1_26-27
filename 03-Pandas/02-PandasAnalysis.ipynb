# Intro to Pandas Analysis

```python
# Import in your libraries:
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'Make': ['Toyota', 'Honda', 'Ford', 'Chevrolet', 'BMW'],
    'Model': ['Corolla', 'Civic', 'Mustang', 'Camaro', '3 Series'],
    'Year': [2010, 2012, 2015, 2018, 2020],
    'Mileage': [120000, 90000, 50000, 30000, 15000]
}
```

## Display Dataframe Data

```python
# Converts our dictionary above into a dataframe:
df = pd.DataFrame(data)

print(df)
```
