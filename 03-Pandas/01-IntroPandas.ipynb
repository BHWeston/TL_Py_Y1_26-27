# Intro to Pandas

```python
# Library Imports:
import pandas as pd
import matplotlib.pyplot as plt

# Python Data Analysis Library
```

```python
# This is a standard list in Python
applesList = [3,2,0,1]
print(applesList)
```

```python
# Let's convert this list to a Series:
applesSeries = pd.Series(applesList)
print(applesSeries)
```

```python
print(applesSeries[1])
```

```python
orangeSeries = pd.Series([0,3,7,2])
print(orangeSeries)
```

```python
# Create a dataframe:
df = pd.DataFrame({
    "apples": applesSeries,
    "oranges": orangeSeries
})

print(df)
```

```python
print(df["apples"])
```
