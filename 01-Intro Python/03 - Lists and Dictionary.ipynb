# List and Dictionaries

```python
lPeople = ["Ben", "Fred", "Billy"]
print(lPeople)
```

```python
print(lPeople[0])
```

Rules of list:
- Can contain multiple values 
- The values can be changed anytime
- List is ordered

```python
lPeople[1] = "Bob"
print(lPeople)
```

## Tuples

```python
tLocations[0] = "Ireland"
```

```python
lPeople.append("John")
print(lPeople)
```
