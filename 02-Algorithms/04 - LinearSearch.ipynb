### Linear Search

```python
# Create a list of bad passwords:
lPoorPasswords = ["password", "12345", "letmein", "qwerty", "admin"]
```

```python
sInputtedPassword = input("Enter in the password that you want to check: ")
```

```python
# Linear Algorithm Time:
# This is known as a flag, this will be used to check whether we have completed the algorithm:
bFound = False

for iCount in range(0, len(lPoorPasswords)):
    if sInputtedPassword == lPoorPasswords[iCount]:
        print("Your password is in the list")
        bFound = True
        break
    else:
        print("Password is not in the list")

# Expand the programme with the following:
# 1 - All inputted passwords, must be a min of 8 letters/numbers
# 2 - Must have a capital
# 3 - Must use numbers
```
